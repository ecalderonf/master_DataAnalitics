```python
# Indexación en arrays
lst = [12,20,3,45,6,23,81,93]
# Seleccionamos desde el elemento 3 (incluido) hasta el final
print(lst[3:])  # Output: [45, 6, 23, 81, 93]

# Seleccionamos desde el elemento 3 hasta el 5. 
# ⚠️ El límite superior del valor que especificamos NO SE INCLUYE.

print(lst[3:6]) # Output: [45, 6, 23]

# Seleccionamos los 4 primeros elementos
lst[:4] # Output: [12, 20, 3, 45]

# Seleccionamos todos los elementos que están en posición impar
print(lst[::2]) # Output: [12, 3, 6, 81]
```

```python
bidimensional = np.random.rand(5, 4)
print(bidimensional)
```

```python
# Accedemos a la fila 0
print(bidimensional[0]) # Output: array([0.63019968, 0.1409182 , 0.930998  , 0.88841414])

# Accedemos al elemento de la fila 0 columna 0
print(bidimensional[0,0]) # Output:  0.6301996771602075

# Esto es lo mismo que lo anterior
print(bidimensional[0][0]) # Output: 0.6301996771602075

# Segunda fila, tercer columna.
print(bidimensional[2,3]) # Output: 0.6651123761198269
```

```python
# Seleccionar las dos primeras filas,y todas las columnas (no ponemos nada después de la coma).
bidimensional[:2,:] # Output: array([[0.63019968, 0.1409182 , 0.930998  , 0.88841414],[0.01226318, 0.19209121, 0.86734263, 0.07889421]])
```

```python
for columnas in range(bidimensional.shape[0]):
    for filas in range(bidimensional.shape[1]):
        if bidimensional[columnas,filas] > 0.5:
            bidimensional[columnas,filas] = -1

print(bidimensional)
```

# Operaciones estadísticas

```python
d = np.random.randint(1,20, (4,5))
print(d)
```

```python
# media
media = np.mean(d)
print(media)
```

```python
# media por filas
media_filas = np.mean(d, axis = 1)
media_filas

# Output: array([8.8, 9.6, 8.8, 9.8])
```

```python
# media por columnas
media_columnas = np.mean(d, axis = 0)
media_columnas

# Output: array([10.  ,  6.25, 12.5 ,  6.  , 11.5 ])
```

```python
# Desviación estándar
np.std(d)

# Output: 5.92347026665957
```

```python
# Desviación estándar por columnas
np.std(d, axis = 0)

# Output: array([4.94974747, 5.67340286, 4.38748219, 4.84767986, 6.34428877])
```

```python
# Desviación estándar por filas

np.std(d, axis = 1)

# Output: array([6.7349833 , 5.95315043, 5.97996656, 4.79165942])
```

```python
# variaza

np.var(d)
# Output: 35.0875
```

```python
# variaza por columnas

np.var(d, axis = 0)

# Output: array([24.5   , 32.1875, 19.25  , 23.5   , 40.25  ])
```

```python
# variaza por filas

np.var(d, axis = 1)

# Output: array([45.36, 35.44, 35.76, 22.96])
```

```python
# Operaciones de comparación

d = np.random.randint(1,20, (4,5))
d
```

```python
print(np.any(d > 10))
# Output: True

print(np.any(d > 20))
# Output: False

print(np.any(d > 10, axis = 0))
# Output: array([ True,  True,  True,  True,  True])

print(np.any(d < 2 , axis = 1 ))
# Output: array([False,  True,  True, False])
```

```python
print(np.all(d > 10))
# Output: True

print(np.all(d > 20))
# Output: False

print(np.all(d > 10, axis = 0))
# Output: array([ True,  True,  True,  True,  True])

print(np.all(d < 2 , axis = 1 ))
# Output: array([False,  True,  True, False])
```
