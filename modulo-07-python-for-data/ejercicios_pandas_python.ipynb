```python
# Importamos las librerías que necesitamos
import pandas as pd
import numpy as np

pd.set_option('display.max_columns',None)

user = pd.read_csv('./data/DatosUsuariosRedesSociales.csv')

user.sample(5)
```

```python
def categorizar_edad (numero):

    if 18 <= numero <= 28:
        return 'Jóven adulto'
    elif 29 <= numero <= 39:
        return 'Adulto jóven'
    elif 40 <= numero <= 50:
        return 'Mediana edad'
    else:
        return 'Adultos mayores'
```

```python
user['age_category'] = user['age'].apply(categorizar_edad)
user
```

```python
user['income'].dtype
```

```python
#Definimos la función
def cambiar_float(numero):
    """
    Función para cambiar el tipo de un número a float.

    Parámetros:
    - numero (int): Número que se va a convertir.

    Retorna:
    - float: El número proporcionado con el tipo cambiado a float.
    """
    return float(numero)

#Llamamos a la función
user['income'] = user['income'].apply(cambiar_float)

#Comprobamos
user['income'].dtype
```

```python
#Vamos a hacer lo mismo para la columna 'time_spent'
user['time_spent'] = user['time_spent'].apply(lambda x: float(x))

user['time_spent'].dtype
```

```python
user['Owns_Car'] = user['Owns_Car'].apply(lambda x: 'yes' if x == True else 'no')
user
```

```python
def travel_female(col_interests,col_gender):
    """
    Función para identificar un usuario con unas características concretas.

    Parámetros:
    - col_interests (str): Columna con los intereses del usuario.

    - col_gender (str): Columna con el género del usuario

    Retorna:
    - bool: Si las condiciones se cumplen o no
    """
    if col_interests == 'Travel' and col_gender == 'female':
        return True
    else:
        return False
```

```python
user['target_audiences']=user.apply(lambda x: travel_female(x['interests'],x['gender']),axis=1)
user
```

```python
def error_edad (edad, num):
    """
    Función corregir la edad introducida en nuestros datos.

    Parámetros:
    - edad (int): Columna con la edad errónea del usuario.

    - num (int): Cantidad de años a sumar

    Retorna:
    - int: Edad corregida
    """
    return (edad + num)
```

```python
user['age']=user['age'].apply(error_edad,num=10)
user
```

```python
def saludo():
    """
    Función que devuelve un saludo cuando la llamamos.

    Parámetros:

    Retorna:
    - str: La frase de saludo
    """
    return 'Hola, buenos días'
```

    - `right` : Valores comunes de ambos DataFrames y no comunes del derecho.
    - `outer` : Utiliza TODOS los valores de ambos DataFrames.
- on: Especifica las columnas comunes en las que se van a basar la unión. Estas columnas DEBEN TENER EL MISMO NOMBRE.
- left_on: Nombre de la columna del df_left en la que se basará la unión.
- right_on: Nombre de la columna del df_right en la que se basará la unión.
'''
