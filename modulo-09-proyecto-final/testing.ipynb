```python
# Testing

import pandas as pd
import os

# Convierte un fichero CSV a JSON y lo guarda en la misma ruta
def convertir_csv_a_json(ruta_csv: str):
    """
    Carga un CSV, lo convierte a JSON y guarda el nuevo fichero
    en la misma ruta con extensión .json.
    """

    # Leer CSV
    df = pd.read_csv(ruta_csv)

    # Construir ruta de salida
    ruta_json = os.path.splitext(ruta_csv)[0] + ".json"

    # Guardar JSON
    df.to_json(ruta_json, orient="records", indent=4)

    print(f"\nCSV convertido a JSON:")
    print(f"- Origen: {ruta_csv}")
    print(f"- Destino: {ruta_json}")
```

```python
convertir_csv_a_json('data/products.csv')
```
